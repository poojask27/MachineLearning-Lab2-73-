```python


def find_s(training_data, target_col="Scholarship", positive_label="Yes"):

    # Get all attributes except the target column
    attributes = [
        attribute for attribute in training_data[0]
        if attribute != target_col
    ]

    # Start with the most specific hypothesis
    hypothesis = {attribute: "0" for attribute in attributes}

    print("Initial Hypothesis:")
    print(hypothesis)
    print("-" * 50)

    # Process each training example
    for i, example in enumerate(training_data, start=1):

        print(f"\nExample {i}: {example}")

        # FIND-S considers only positive examples
        if example[target_col] == positive_label:

            print("Positive Example -> Updating Hypothesis")

            for attribute in attributes:

                # First positive example
                if hypothesis[attribute] == "0":
                    hypothesis[attribute] = example[attribute]

                # If values differ, generalize
                elif hypothesis[attribute] != example[attribute]:
                    hypothesis[attribute] = "?"

            print("Current Hypothesis:")
            print(hypothesis)

        else:
            print("Negative Example -> Ignored by FIND-S")

    return hypothesis


# --------------------------------------------------
# Dataset: Student Scholarship Eligibility
# Total Records = 20
# --------------------------------------------------

training_data = [

    # 1
    {
        "Attendance": "High",
        "Marks": "Excellent",
        "Projects": "Completed",
        "FamilyIncome": "Low",
        "Scholarship": "Yes"
    },

    # 2
    {
        "Attendance": "High",
        "Marks": "Excellent",
        "Projects": "Completed",
        "FamilyIncome": "Medium",
        "Scholarship": "Yes"
    },

    # 3
    {
        "Attendance": "Low",
        "Marks": "Average",
        "Projects": "Incomplete",
        "FamilyIncome": "Low",
        "Scholarship": "No"
    },

    # 4
    {
        "Attendance": "High",
        "Marks": "Excellent",
        "Projects": "Completed",
        "FamilyIncome": "High",
        "Scholarship": "Yes"
    },

    # 5
    {
        "Attendance": "Medium",
        "Marks": "Good",
        "Projects": "Completed",
        "FamilyIncome": "Low",
        "Scholarship": "No"
    },

    # 6
    {
        "Attendance": "High",
        "Marks": "Excellent",
        "Projects": "Completed",
        "FamilyIncome": "Low",
        "Scholarship": "Yes"
    },

    # 7
    {
        "Attendance": "Low",
        "Marks": "Poor",
        "Projects": "Incomplete",
        "FamilyIncome": "High",
        "Scholarship": "No"
    },

    # 8
    {
        "Attendance": "High",
        "Marks": "Excellent",
        "Projects": "Completed",
        "FamilyIncome": "Medium",
        "Scholarship": "Yes"
    },

    # 9
    {
        "Attendance": "Medium",
        "Marks": "Average",
        "Projects": "Incomplete",
        "FamilyIncome": "Medium",
        "Scholarship": "No"
    },

    # 10
    {
        "Attendance": "High",
        "Marks": "Excellent",
        "Projects": "Completed",
        "FamilyIncome": "High",
        "Scholarship": "Yes"
    },

    # 11
    {
        "Attendance": "Low",
        "Marks": "Good",
        "Projects": "Incomplete",
        "FamilyIncome": "Low",
        "Scholarship": "No"
    },

    # 12
    {
        "Attendance": "High",
        "Marks": "Excellent",
        "Projects": "Completed",
        "FamilyIncome": "Low",
        "Scholarship": "Yes"
    },

    # 13
    {
        "Attendance": "Medium",
        "Marks": "Good",
        "Projects": "Incomplete",
        "FamilyIncome": "High",
        "Scholarship": "No"
    },

    # 14
    {
        "Attendance": "High",
        "Marks": "Excellent",
        "Projects": "Completed",
        "FamilyIncome": "Medium",
        "Scholarship": "Yes"
    },

    # 15
    {
        "Attendance": "Low",
        "Marks": "Average",
        "Projects": "Completed",
        "FamilyIncome": "High",
        "Scholarship": "No"
    },

    # 16
    {
        "Attendance": "High",
        "Marks": "Excellent",
        "Projects": "Completed",
        "FamilyIncome": "High",
        "Scholarship": "Yes"
    },

    # 17
    {
        "Attendance": "Medium",
        "Marks": "Poor",
        "Projects": "Incomplete",
        "FamilyIncome": "Low",
        "Scholarship": "No"
    },

    # 18
    {
        "Attendance": "High",
        "Marks": "Excellent",
        "Projects": "Completed",
        "FamilyIncome": "Low",
        "Scholarship": "Yes"
    },

    # 19
    {
        "Attendance": "Low",
        "Marks": "Good",
        "Projects": "Incomplete",
        "FamilyIncome": "Medium",
        "Scholarship": "No"
    },

    # 20
    {
        "Attendance": "High",
        "Marks": "Excellent",
        "Projects": "Completed",
        "FamilyIncome": "Medium",
        "Scholarship": "Yes"
    }
]


# Run the FIND-S Algorithm
final_hypothesis = find_s(training_data)


print("\n" + "=" * 50)
print("FINAL HYPOTHESIS")
print("=" * 50)
print(final_hypothesis)

```

```python

# ============================================================
# DATA VISUALIZATION
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


# Convert existing training_data into DataFrame
df = pd.DataFrame(training_data)


# ------------------------------------------------------------
# GRAPH 1: Scholarship Distribution
# ------------------------------------------------------------

scholarship_count = df["Scholarship"].value_counts()

plt.figure(figsize=(6, 4))

plt.bar(
    scholarship_count.index,
    scholarship_count.values
)

plt.title("Scholarship Eligibility Distribution")
plt.xlabel("Scholarship Status")
plt.ylabel("Number of Students")

plt.show()


# ------------------------------------------------------------
# GRAPH 2: Attendance vs Scholarship
# ------------------------------------------------------------

attendance_graph = pd.crosstab(
    df["Attendance"],
    df["Scholarship"]
)

attendance_graph.plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Attendance vs Scholarship")
plt.xlabel("Attendance")
plt.ylabel("Number of Students")

plt.xticks(rotation=0)
plt.legend(title="Scholarship")

plt.show()


# ------------------------------------------------------------
# GRAPH 3: Marks vs Scholarship
# ------------------------------------------------------------

marks_graph = pd.crosstab(
    df["Marks"],
    df["Scholarship"]
)

marks_graph.plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Marks vs Scholarship")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.xticks(rotation=0)
plt.legend(title="Scholarship")

plt.show()


# ------------------------------------------------------------
# GRAPH 4: Projects vs Scholarship
# ------------------------------------------------------------

projects_graph = pd.crosstab(
    df["Projects"],
    df["Scholarship"]
)

projects_graph.plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Projects vs Scholarship")
plt.xlabel("Project Status")
plt.ylabel("Number of Students")

plt.xticks(rotation=0)
plt.legend(title="Scholarship")

plt.show()


# ------------------------------------------------------------
# GRAPH 5: Family Income vs Scholarship
# ------------------------------------------------------------

income_graph = pd.crosstab(
    df["FamilyIncome"],
    df["Scholarship"]
)

income_graph.plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Family Income vs Scholarship")
plt.xlabel("Family Income")
plt.ylabel("Number of Students")

plt.xticks(rotation=0)
plt.legend(title="Scholarship")

plt.show()

```
